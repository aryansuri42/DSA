```python
# 3x3 Matrix traverse the target Count the number of ways

# Return how many paths

def count_paths(r,c,path):
  if r==1 and c==1:
    print(path)
  if r>1:
    count_paths(r-1,c,path+"D")
  if c>1:
    count_paths(r,c-1,path+"R")
```

```python
count_paths(3,3,"")
```

```python
def count_paths_diagonal(r,c,path):
  if r==1 and c==1:
    print(path)
  if r>1:
    count_paths_diagonal(r-1,c,path+"D")
  if c>1:
    count_paths_diagonal(r,c-1, path+"R")
  if r>1 and c>1:
    count_paths_diagonal(r-1,c-1,path+"X")
```

```python
## Maze with obstacle
def count_paths_obstacle(r,c,path,maze):
  if r==len(maze)-1 and c==len(maze[0])-1:
    print(path)
    return
  if maze[r][c]==False:
    return
  # i am considering this block in my path
  maze[r][c] = False
  if r<len(maze)-1:
    count_paths_obstacle(r+1,c,path+"D", maze)
  if c<len(maze[0])-1:
    count_paths_obstacle(r,c+1,path+"R", maze)
  if r>0:
    count_paths_obstacle(r-1,c,path+"U", maze)
  if c>0:
    count_paths_obstacle(r,c-1,path+"L", maze)
  # this line is where the function is over
  # so before the function gets removed also remove the changes that were made by that function
  maze[r][c] = True
count_paths_obstacle(0,0,"",[[True,True,True],[True,True,True],[True,True,True]])

# Make a change and reverse the change
```

```python
# Print the path
def count_steps(r,c,count,maze):
  if r==len(maze)-1 and c==len(maze[0])-1:
    print(count)
    return

  if maze[r][c]==False:
    return

  maze[r][c]=False

  if r<len(maze)-1:
    count_steps(r+1,c,count+1,maze)
  if c<len(maze[0])-1:
    count_steps(r,c+1,count+1,maze)
  if r>0:
    count_steps(r-1,c,count+1,maze)
  if c>0:
    count_steps(r,c-1,count+1,maze)
  maze[r][c]=True
count_steps(0,0,1,[[True,True,True],[True,True,True],[True,True,True]])
```
