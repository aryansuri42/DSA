```python
# Print the path
def count_steps(r,c,count,maze):
  if r==len(maze)-1 and c==len(maze[0])-1:
    print(count)
    return

  if maze[r][c]==False:
    return

  maze[r][c]=False

  if r<len(maze)-1:
    count_steps(r+1,c,count+1,maze)
  if c<len(maze[0])-1:
    count_steps(r,c+1,count+1,maze)
  if r>0:
    count_steps(r-1,c,count+1,maze)
  if c>0:
    count_steps(r,c-1,count+1,maze)
  maze[r][c]=True
count_steps(0,0,1,[[True,True,True],[True,True,True],[True,True,True]])
```

```python
def queens(board, row):
    if row == len(board):
        display(board)
        print()  # For better separation of solutions
        return 1

    count = 0
    for col in range(len(board)):
        if isSafe(board, row, col):
            board[row][col] = True
            count += queens(board, row + 1)
            board[row][col] = False  # Backtracking

    return count  # Ensure count is returned

def isSafe(board, row, col):
    # Check vertical row
    for i in range(row):
        if board[i][col]:  # No need to compare with True
            return False

    # Check diagonal left
    maxLeft = min(row, col)
    for i in range(1, maxLeft + 1):
        if board[row - i][col - i]:
            return False

    # Check diagonal right
    maxRight = min(row, len(board) - col - 1)
    for i in range(1, maxRight + 1):
        if board[row - i][col + i]:
            return False

    return True

def display(board):
    for i in board:
        for j in i:
            print("Q" if j else "_", end=" ")
        print()

# Dynamically create an NxN board
N = 4  # Change N for different board sizes
board = [[False] * N for _ in range(N)]

# Start solving and count solutions
solution_count = queens(board, 0)
print(f"Total solutions: {solution_count}")
```
